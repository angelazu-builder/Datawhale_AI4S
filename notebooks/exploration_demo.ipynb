# Extended Kesten Left-Tail Wealth Dynamics Research
## Phase Transition Discovery & Agent Exploration Notebook

This notebook demonstrates:
1. **Single-Seed Kesten Simulator**: Vectorized wealth distribution simulation with boundary condition handling.
2. **No-Intervention & Random Baseline**: Reference baseline evaluations.
3. **Adaptive Agent Active Search**: 120-round exploration identifying phase transition threshold $R^* = p/c$.
4. **Visual Analytics**: Publication-quality plots of phase transitions and left-tail power law fits.

```python
import sys
import os
sys.path.insert(0, os.path.abspath('..'))
import matplotlib.pyplot as plt

from src import (
    SimulationConfig,
    KestenSimulator,
    ExperimentLogger,
    BaselineAgent,
    RandomAgent,
    AdaptiveAgent,
    Visualizer
)

# Initialize configuration
config = SimulationConfig()
simulator = KestenSimulator(config)
visualizer = Visualizer(config.output_dir)
print(f"Simulator ready: Population N={config.population_size}, Steps T={config.time_steps}")
```

### 1. Single Simulation Run
Run a single trial with $p=0.01$, $c=0.10$, `reflect` boundary, and `lognormal` income distribution across 5 seeds.

```python
single_res = simulator.run_trial(p=0.01, c=0.10, boundary_type='reflect', income_dist='lognormal')
print(f"Beta Left Exponent: {single_res['beta_left']:.4f} ± {single_res['beta_ci']:.4f}")
print(f"Poverty Rate (< 1.0): {single_res['poverty_rate']*100:.2f}%")
print(f"Kurtosis: {single_res['kurtosis']:.4f}")
```

### 2. Run Baseline Evaluations

```python
b_agent = BaselineAgent(simulator, config)
base_res = b_agent.run_baseline()
print(f"No-Intervention Baseline Beta Left: {base_res['beta_left_base']:.4f}")
```

### 3. Run Adaptive Active Discovery Agent (120 Rounds)

```python
logger = ExperimentLogger()
a_agent = AdaptiveAgent(simulator, config, logger)
history, mutations = a_agent.run_exploration()
print(f"Exploration completed with {len(mutations)} phase transition signals detected.")
```

### 4. Visualize Phase Transition Diagram $R^* = p/c$

```python
fig1 = visualizer.plot_phase_transition(history)
fig2 = visualizer.plot_agent_trajectory(history)
print("Visualizations generated successfully.")
```
